```python
from datetime import date, datetime, timedelta
import json
from bs4 import BeautifulSoup
import sys
from pathlib import Path

# Adiciona o diretório raiz do projeto ao sys.path
sys.path.append(str(Path.cwd().parent))

from pacote.functions import *
```

```python
url_calendar_list = ['https://www.b3.com.br/pt_br/noticias/calendario-de-feriados-2025.htm']
teste_calendar = web_scraping_f(url_calendar_list)
```

```python
meses = {
    "janeiro": 1, "fevereiro": 2, "março": 3, "abril": 4,
    "maio": 5, "junho": 6, "julho": 7, "agosto": 8,
    "setembro": 9, "outubro": 10, "novembro": 11, "dezembro": 12
}
soup_calendar = BeautifulSoup(teste_calendar[0], 'html.parser')
list_table = soup_calendar.find_all('tbody')
list_tbody = list_table[0].find_all('p') + list_table[1].find_all('p')
list_dates = []
for i in range(0, len(list_tbody), 2):
    date_splited = list_tbody[i].text.split(' de ')
    date_str = [2025, meses.get(date_splited[1]), int(date_splited[0])]
#     list_dates.append(date(*date_str))

    list_dates.append(date_str)
```

```python
salva_dict(list_dates, "../arquivos/json/2025_calendar.json")
```
